```python
# Step 1: Set Up Your Environment
from pyspark.sql import SparkSession
```

```python
# Initialize Spark session
spark = SparkSession.builder.appName("StudentGradesRegression").getOrCreate()
```

```python
# Step 2: Load the Data
data = spark.read.csv("Student_Grades_Data.csv", header=True, inferSchema=True)
data.show(3)
```

```python
# Step 3: Data Preprocessing
from pyspark.ml.feature import StringIndexer, VectorAssembler, StandardScaler
from pyspark.ml import Pipeline
```

```python
# Example: Assembling feature columns into a feature vector
assembler = VectorAssembler(inputCols=["Time_to_Study"], outputCol="features")
```

```python
# Step 4: Apply Feature Scaling
scaler = StandardScaler(inputCol="features", outputCol="scaled_features", withMean=True, withStd=True)
```

```python
# Create pipeline for preprocessing
pipeline = Pipeline(stages=[assembler, scaler])
data_preprocessed = pipeline.fit(data).transform(data)
```

```python
# Split the data into training and testing sets
train_data, test_data = data_preprocessed.randomSplit([0.8, 0.2], seed=1)
```

```python
# Step 4: Build the Regression Model
from pyspark.ml.regression import LinearRegression
# Initialize the regressor
lr = LinearRegression(featuresCol="features", labelCol="Grades")
```

```python
# Train the model
lr_model = lr.fit(train_data)
```

```python
# Step 5: Evaluate the Model
from pyspark.ml.evaluation import RegressionEvaluator
# Make predictions
predictions = lr_model.transform(test_data)
```

```python
# Evaluate the model
evaluator = RegressionEvaluator(labelCol="Grades", predictionCol="prediction",
metricName="rmse")
rmse = evaluator.evaluate(predictions)
print(f"Test RMSE = {rmse}")
```

```python
# Step 6: Save the Model (Optional)
lr_model.save("path/to/save/model")
```
